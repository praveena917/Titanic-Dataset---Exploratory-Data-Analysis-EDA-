<h4>Importing Libraries</h4>

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
```

```python
df = pd.read_csv("Titanic-Dataset.csv")

df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
#checking for missing values
df.isnull().sum()
```

<h4>Visualizations</h4>

<h4>Survival Count</h4>

```python
sns.countplot(x='Survived', data=df)
plt.title("Survival Distribution")
plt.savefig("survival_count.png",dpi=300,bbox_inches='tight')
plt.show()
```

<h4>Survival by Gender</h4>

```python
sns.countplot(x='Sex', hue='Survived', data=df)
plt.title("Survival by Gender")
plt.savefig("survival_gender.png",dpi=300,bbox_inches="tight")
plt.show()
```

<h4>Survival by Passenger Class</h4>

```python
sns.countplot(x='Pclass', hue='Survived', data=df)
plt.title("Survival by Passenger Class")
plt.savefig("survival_passenger.png",dpi=300,bbox_inches="tight")
plt.show()
```

<h4>Histogram of Age</h4>

```python
plt.figure(figsize=(8,5))
sns.histplot(df['Age'], bins=30, kde=True)
plt.title("Age Distribution")
plt.savefig("age.png",dpi=300,bbox_inches="tight")
plt.show()
```

<h4>Boxplot of Fare</h4>

```python
plt.figure(figsize=(8,5))
sns.boxplot(x=df['Fare'])
plt.title("Fare Distribution")
plt.savefig("Fare.png",dpi=300,bbox_inches="tight")
plt.show()
```

<h4>Correlation of Heatmap</h4>

```python
df2 = df.copy()

df2['Sex'] = df2['Sex'].map({'male':0,'female':1})

corr = df2.corr(numeric_only=True)

plt.figure(figsize=(10,8))
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.savefig("correlation.png",dpi=300,bbox_inches="tight")
plt.show()
```

<h4>Pairplot</h4>

```python
sns.pairplot(df[['Age','Fare','Pclass','Survived']])
plt.savefig("pairplot.png",dpi=300,bbox_inches="tight")
plt.show()
```
